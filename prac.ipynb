```python
P_Python = 0.2
P_Java = 0.8
P_Placed_given_Python = 0.9
P_Placed_given_Java = 0.4

P_Placed = (P_Python * P_Placed_given_Python) + (P_Java * P_Placed_given_Java)
P_Python_given_Placed = (P_Python * P_Placed_given_Python) / P_Placed
P_Java_given_Placed = (P_Java * P_Placed_given_Java) / P_Placed

print(f"Probability student knows Python: {P_Python:.2f}")
print(f"Probability student knows Java: {P_Java:.2f}")

print(f"Probability placed student knows Python: {P_Python_given_Placed:.2f}")

```

```python
P_Green = 0.85
P_Blue = 0.15
P_WitnessBlue_given_Blue = 0.8
P_WitnessBlue_given_Green = 0.2

P_WitnessBlue = (P_Blue * P_WitnessBlue_given_Blue) + (P_Green * P_WitnessBlue_given_Green)
P_Blue_given_WitnessBlue = (P_Blue * P_WitnessBlue_given_Blue) / P_WitnessBlue

print(f"Probability taxi is actually Blue: {P_Blue_given_WitnessBlue:.3f}")

```
